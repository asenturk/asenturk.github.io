## 4. İndeksleme ve dilimleme


````python
dizi = np.arange(0, 11)

print(dizi[3])
print(dizi[-1])
print(dizi[2:6])
print(dizi[::2])
````

İki boyutlu dizilerde `[satır, sütun]` kullanımı tercih edilir:

### Görünüm ve kopya farkı

NumPy dilimleri çoğunlukla orijinal veriyi paylaşan bir görünüm oluşturur:


````python
dizi = np.arange(10)
parca = dizi[0:4]
parca[:] = 99
print(dizi)
````

````python
dizi = np.arange(10)
parca = dizi[0:4].copy()
parca[:] = 99
print(dizi)
````

## 6. Vektörleştirilmiş işlemler

Döngü yazmadan tüm elemanlara işlem uygulanabilir:


````python
dizi = np.arange(1, 6)

print(dizi + 10)
print(dizi * 2)
print(dizi ** 2)
print(np.sqrt(dizi))
print(np.sin(dizi))
````

Eleman eleman işlemler için iki dizinin şekilleri uyumlu olmalıdır:

## 7. Broadcasting

NumPy, uyumlu şekillerdeki küçük diziyi daha büyük dizi boyunca genişletir:


````python
matris = np.array([[1, 2, 3],
                   [4, 5, 6]])
satir = np.array([10, 20, 30])

print(matris + satir)
````

Her sütunun ortalamasını çıkarma:


````python
ortalama = matris.mean(axis=0)
merkezlenmis = matris - ortalama
print(merkezlenmis)
````

## 8. Boolean filtreleme


````python
olcumler = np.array([12.1, 9.8, 15.4, 7.2, 11.5])
maske = olcumler > 10

print(maske)
print(olcumler[maske])
print(olcumler[olcumler > 10])
````

Birden fazla koşulda her koşul paranteze alınır ve `&`, `|`, `~` kullanılır:


````python
print(olcumler[(olcumler >= 10) & (olcumler <= 13)])
````

## 9. Özetleme ve `axis`


````python
matris = np.array([[1, 2, 3],
                   [4, 5, 6]])

print(matris.sum())
print(matris.sum(axis=0))  # satırlar boyunca: sütun sonuçları
print(matris.sum(axis=1))  # sütunlar boyunca: satır sonuçları
print(matris.mean(axis=0))
print(matris.min(), matris.argmin())
print(matris.max(), matris.argmax())
````

Mühendislik verilerinde sık kullanılan diğer fonksiyonlar:


````python
print(np.median(matris))
print(np.std(matris))
````

## 10. Matris çarpımı ve doğrusal cebir

`*` eleman eleman, `@` matris çarpımı yapar.


````python
a = np.array([[1, 2, 3], [4, 5, 6]])
b = np.array([[1, 2], [3, 4], [5, 6]])

print(a @ b)
````

Doğrusal denklem sistemi `Ax = y`:


````python
A = np.array([[1, 1, 1],
              [2, 1, 2],
              [1, 2, 3]], dtype=float)
y = np.array([9, 15, 20], dtype=float)

determinant = np.linalg.det(A)
x = np.linalg.solve(A, y)

print(determinant)
print(x)
print(A @ x)
````

Denklem çözmek için ters matris hesaplamak yerine `np.linalg.solve` tercih edilir. Determinant sıfıra çok yakınsa sistem tekil veya sayısal olarak sorunlu olabilir.

## 11. Bütünleştirici uygulama: Sensör verisi


````python
rng = np.random.default_rng(7)
olcumler = rng.normal(loc=25, scale=2, size=(5, 8))

gunluk_ortalama = olcumler.mean(axis=1)
anormal_maske = (olcumler < 21) | (olcumler > 29)

print(gunluk_ortalama)
print(olcumler[anormal_maske])
````
