## 3. Aritmetik operatörler


```python
a, b = 23, 5

print(a + b)   # toplama
print(a - b)   # çıkarma
print(a * b)   # çarpma
print(a / b)   # gerçek bölme
print(a // b)  # tam bölme
print(a % b)   # kalan
print(a ** b)  # üs alma
```

İşlem önceliğini görünür kılmak için parantez kullanılmalıdır.


```python
yaricap = 4
pi = 3.14159
alan = pi * yaricap ** 2
print(alan)
```

## 5. String türü


```python
metin = "Bu bir metindir."

print(len(metin))
print(metin.upper())
print(metin.lower())
print(metin.count("i"))
print(metin.replace("metin", "örnek"))
print(metin.split())
```

Stringler indekslenebilir. İlk elemanın indeksi `0`, son elemanın indeksi `-1`'dir. Dilimlemede bitiş indeksi sonuca dahil edilmez.


```python
metin = "Bu bir metindir."

print(metin[0])
print(metin[-1])
print(metin[0:3])
print(metin[:5])
print(metin[5:])
print(metin[::2])
print(metin[::-1])
```

String birleştirme ve tekrar:


```python
birinci_kelime = "Merhaba"
ikinci_kelime = "Dünya"

print(birinci_kelime + " " + ikinci_kelime)
print("-" * 30)
```

Stringler değiştirilemez (immutable). Aşağıdaki atama hata verir:


```python
metin = "Python"
# metin[0] = "p"
```

## 6. Klavyeden veri alma ve tür dönüşümü

`input` her zaman `str` döndürür.


```python
birinci = input("Birinci sayıyı giriniz: ")
ikinci = input("İkinci sayıyı giriniz: ")

toplam = float(birinci) + float(ikinci)
print(f"Toplam: {toplam}")
```

### Uygulama: Silindirin hacmi


```python
import math

yaricap = float(input("Yarıçapı giriniz: "))
yukseklik = float(input("Yüksekliği giriniz: "))
hacim = math.pi * yaricap ** 2 * yukseklik

print(f"Silindirin hacmi: {hacim:.2f}")
```

### Uygulama: Metindeki boşluk sayısı


```python
metin = input("Bir metin giriniz: ")
bosluk_sayisi = metin.count(" ")
print(f"Girilen metinde {bosluk_sayisi} boşluk vardır.")
```
